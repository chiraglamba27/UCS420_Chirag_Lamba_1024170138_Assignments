```python
import random

roll_number = "1024170138"

L = [int(d) * 10 for d in roll_number]
print("Q1.i", L)

L.append(95)
print("Q1.ii append", L)

L.insert(4, 65)
print("Q1.ii insert", L)

L.remove(65)
print("Q1.iii remove", L)

L.pop()
print("Q1.iii pop", L)

L.sort()
print("Q1.iv ascending", L)

L.sort(reverse=True)
print("Q1.iv descending", L)

print("Q1.v first three", L[:3])
print("Q1.v last three", L[-3:])

avg_L = sum(L) / len(L)
above_avg_L = [x for x in L if x > avg_L]
print("Q1.vi above average", above_avg_L)


original_L = [int(d) * 10 for d in roll_number]
scores = tuple(original_L[:8])
print("Q2 scores", scores)

highest = max(scores)
highest_index = scores.index(highest)
lowest = min(scores)
lowest_count = scores.count(lowest)
print("Q2.i highest", highest, "index", highest_index)
print("Q2.i lowest", lowest, "count", lowest_count)

reversed_scores = list(reversed(scores))
print("Q2.ii reversed", reversed_scores)

user_score = int(input("Enter a score to search for: "))
if user_score in scores:
    print("Q2.iii found at index", scores.index(user_score))
else:
    print("Q2.iii not present in the tuple")

try:
    scores[0] = 100
except TypeError as e:
    print("Q2.iv error", e)

first_score, second_score, *remaining_scores = scores
print("Q2.v first", first_score, "second", second_score, "rest", remaining_scores)


random.seed(int(roll_number))
numbers = [random.randint(100, 900) for _ in range(100)]
print("Q3.i", numbers)

odd_numbers = [n for n in numbers if n % 2 != 0]
print("Q3.ii odd count", len(odd_numbers))

even_numbers = [n for n in numbers if n % 2 == 0]
print("Q3.iii even count", len(even_numbers))


def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True


prime_numbers = [n for n in numbers if is_prime(n)]
print("Q3.iv prime count", len(prime_numbers))
print("Q3.iv prime list", prime_numbers)

most_common = max(set(numbers), key=numbers.count)
print("Q3.v most frequent", most_common, "occurs", numbers.count(most_common), "times")


digits = [int(d) for d in roll_number[:8]]
A = {d * 7 for d in digits}
B = {d * 9 for d in digits}
print("Q4 set A", A)
print("Q4 set B", B)

print("Q4.vi union", A | B)
print("Q4.vii intersection", A & B)
print("Q4.viii A minus B", A.difference(B))
print("Q4.viii B minus A", B.difference(A))
print("Q4.ix symmetric difference", A.symmetric_difference(B))
print("Q4.x A subset of B", A.issubset(B))
print("Q4.x B superset of A", B.issuperset(A))

X = int(input("Enter a value to discard from set A: "))
A.discard(X)
print("Q4.xi A after discard", A)


my_dict = {
    "name": "Simran Kaur",
    "roll_no": "1024170138",
    "branch": "Computer Science",
    "age": 20,
    "city": "Amritsar"
}

value = my_dict.pop("city")
my_dict["location"] = value
print("Q5.i", my_dict)

my_dict["cgpa"] = 8.4
print("Q5.ii", my_dict)

my_dict["age"] += 1
print("Q5.iii", my_dict)

dict_pop_version = dict(my_dict)
removed_value = dict_pop_version.pop("branch")
print("Q5.iv pop version", dict_pop_version, "removed value", removed_value)

dict_del_version = dict(my_dict)
del dict_del_version["branch"]
print("Q5.iv del version", dict_del_version)

for key, value in my_dict.items():
    print(f"Q5.v {key} \u2192 {value}")

if "email" in my_dict:
    print("Q5.vi", my_dict["email"])
else:
    print("Q5.vi email key not found")

friend_dict = {
    "name": "Rahul Mehta",
    "roll_no": "1024170200",
    "branch": "Electronics",
    "age": 21,
    "city": "Chandigarh"
}

merged_dict = {**my_dict, **friend_dict}
print("Q5.vii", merged_dict)

string_only_dict = {k: v for k, v in my_dict.items() if isinstance(v, str)}
print("Q5.viii", string_only_dict)
```
