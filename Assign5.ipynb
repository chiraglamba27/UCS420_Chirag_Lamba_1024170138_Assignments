```python
import numpy as np

print("Q1. Basic NumPy Array Operations")

arr = np.array([10, 20, 30, 40, 50])

print("Original Array:", arr)
print("Addition of 2:", arr + 2)
print("Multiplication by 3:", arr * 3)
print("Division by 2:", arr / 2)


print("\nQ2. Basic NumPy Array")


arr = np.array([1, 2, 3, 6, 4, 5])

print("Original Array:", arr)
print("Reversed Array:", arr[::-1])


x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

value_x = np.bincount(x).argmax()
indices_x = np.where(x == value_x)[0]

value_y = np.bincount(y).argmax()
indices_y = np.where(y == value_y)[0]

print("\nFor x:")
print("Most frequent value:", value_x)
print("Indices:", indices_x)

print("\nFor y:")
print("Most frequent value:", value_y)
print("Indices:", indices_y)


print("\nQ3. Accessing Elements from 2-D Array")

arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])

print("1st row, 2nd column:", arr[0, 1])
print("3rd row, 1st column:", arr[2, 0])


print("\nQ4. NumPy Array Using linspace()")

Chirag = np.linspace(10, 100, 25)

print("Array:")
print(Chirag)

print("Dimensions:", Chirag.ndim)
print("Shape:", Chirag.shape)
print("Total elements:", Chirag.size)
print("Data type:", Chirag.dtype)
print("Total bytes:", Chirag.nbytes)

transpose_Chirag = Chirag.reshape(1, 25)

print("Transpose using reshape():")
print(transpose_Chirag)

print("Transpose using T:")
print(Chirag.T)

print("For a 1-D array, T does not change the shape.")


print("\nQ5. 2-D Array Operations")

ucs420_Chirag = np.array([[10, 20, 30, 40],
                          [50, 60, 70, 80],
                          [90, 15, 20, 35]])

print("Original Array:")
print(ucs420_Chirag)

print("Mean:", np.mean(ucs420_Chirag))
print("Median:", np.median(ucs420_Chirag))
print("Maximum:", np.max(ucs420_Chirag))
print("Minimum:", np.min(ucs420_Chirag))
print("Unique Elements:", np.unique(ucs420_Chirag))

reshaped_ucs420_Chirag = ucs420_Chirag.reshape(4, 3)

print("\nReshaped Array:")
print(reshaped_ucs420_Chirag)

resized_ucs420_Chirag = np.resize(ucs420_Chirag, (2, 3))

print("\nResized Array:")
print(resized_ucs420_Chirag)
```
