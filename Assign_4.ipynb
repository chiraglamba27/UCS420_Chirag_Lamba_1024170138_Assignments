```python
import pandas as pd

roll_no = input("Enter your college roll number: ")

last_digit = int(roll_no[-1])
second_last_digit = int(roll_no[-2])

categories = ["billing", "account", "general"]

cat1 = categories[second_last_digit % 3]
cat2 = categories[last_digit % 3]

personalized_entries = [
    {
        "question": "how can i check my billing history",
        "answer": "You can check your billing history from the billing section.",
        "keywords": "billing history invoices charges",
        "category": cat1
    },
    {
        "question": "how do i update my account details",
        "answer": "You can update your account details from the account settings.",
        "keywords": "account update details settings",
        "category": cat2
    }
]

fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]

faq_data = fixed_entries + personalized_entries

df = pd.DataFrame(faq_data)

print("\nQ1: Final Knowledge Base")
print(df)


def score_query(query, df):
    query_words = set(query.lower().split())
    results = []

    for i, row in df.iterrows():
        question_words = set(row["question"].lower().split())
        keyword_words = set(row["keywords"].lower().split())

        score = len(query_words & question_words)
        score += len(query_words & keyword_words)

        if score > 0:
            results.append((score, i, row["question"]))

    results.sort(reverse=True)

    print("\nQ2: Matching Entries")

    if len(results) == 0:
        print("No matching entries found.")
    else:
        for score, index, question in results:
            print("Confidence:", score, "|", question)

    return results


query = input("\nEnter a query for Q2: ")
score_query(query, df)


def same_category(category_name, df):
    return df[df["category"] == category_name][["question", "answer", "keywords", "category"]]


personalized_category = personalized_entries[0]["category"]

print("\nQ3: Entries in category:", personalized_category)
print(same_category(personalized_category, df))


print("\nQ4: Add a new keyword")

entry_number = 1
new_keyword = input("Enter a new keyword: ")

old_keywords = df.loc[entry_number, "keywords"]
df.loc[entry_number, "keywords"] = old_keywords + " " + new_keyword

df.to_csv("_faq_data.csv", index=False)

print("\nUpdated DataFrame")
print(df)
print("\nData saved to _faq_data.csv")


print("\nQ5: Number of FAQ entries per category")
print(df.groupby("category").size())


print("\nQ6: Tie Demonstration")

print("\nQuery producing a tie:")
tie_results = score_query("fee", df)

if len(tie_results) >= 2:
    highest_score = tie_results[0][0]
    highest_matches = [x for x in tie_results if x[0] == highest_score]

    print("\nHighest scoring entries:")
    for score, index, question in highest_matches:
        print("Confidence:", score, "|", question)

print("\nQuery not producing a tie:")
score_query("password", df)
```
