```python

import re
import nltk
import numpy as np
import matplotlib.pyplot as plt

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist
from nltk.stem import PorterStemmer, WordNetLemmatizer

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

from textblob import TextBlob
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
from wordcloud import WordCloud

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense


# Q1

paragraph = """
Technology has changed the way people live and work.
Computers and smartphones make communication faster and easier.
Artificial intelligence helps people solve complex problems.
Students use technology to learn new skills and explore ideas.
Technology continues to create new opportunities in education and business.
"""

clean_text = re.sub(r'[^\w\s]', '', paragraph.lower())

print("\nQ1")

print("\nLowercase text:")
print(clean_text)

words = word_tokenize(clean_text)
sentences = sent_tokenize(paragraph)

print("\nWord Tokenization:")
print(words)

print("\nSentence Tokenization:")
print(sentences)

print("\nPython split():")
print(clean_text.split())

print("\nNLTK word_tokenize():")
print(word_tokenize(clean_text))

stop_words = set(stopwords.words('english'))

filtered_words = [word for word in words if word not in stop_words]

print("\nAfter removing stopwords:")
print(filtered_words)

freq = FreqDist(filtered_words)

print("\nWord Frequency:")
print(freq.most_common())


# Q2

print("\nQ2")

alphabet_words = re.findall(r'\b[a-zA-Z]+\b', paragraph.lower())

filtered_words = [word for word in alphabet_words if word not in stop_words]

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

stemmed_words = [stemmer.stem(word) for word in filtered_words]
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]

print("\nAlphabetic Words:")
print(alphabet_words)

print("\nAfter Stopword Removal:")
print(filtered_words)

print("\nStemmed Words:")
print(stemmed_words)

print("\nLemmatized Words:")
print(lemmatized_words)

print("\nComparison:")
print("Stemming removes word endings without considering meaning.")
print("Lemmatization converts words into meaningful dictionary forms.")
print("Lemmatization is preferred when meaningful words are important.")


# Q3

print("\nQ3")

texts = [
    "Artificial intelligence is changing the world of technology.",
    "Cricket is a popular sport played by millions of people.",
    "Online shopping makes buying products easy and convenient."
]

count_vectorizer = CountVectorizer()
bow = count_vectorizer.fit_transform(texts)

print("\nBag of Words:")
print(bow.toarray())
print(count_vectorizer.get_feature_names_out())

tfidf_vectorizer = TfidfVectorizer()
tfidf = tfidf_vectorizer.fit_transform(texts)

print("\nTF-IDF Scores:")
print(tfidf.toarray())

features = tfidf_vectorizer.get_feature_names_out()

for i, text in enumerate(texts):
    scores = tfidf[i].toarray()[0]
    top_words = scores.argsort()[-3:][::-1]

    print("\nText:", text)
    print("Top 3 Keywords:")

    for index in top_words:
        print(features[index], round(scores[index], 3))


# Q4

print("\nQ4")

ai_text = """
Artificial intelligence allows machines to learn from data.
AI is used in healthcare, education and business.
It helps automate tasks and solve complex problems.
Machine learning is an important part of artificial intelligence.
"""

blockchain_text = """
Blockchain is a technology used to store digital information.
It provides secure and transparent transactions.
Blockchain is commonly used in cryptocurrencies.
It removes the need for a central authority.
"""

ai_words = set(re.findall(r'\b[a-zA-Z]+\b', ai_text.lower()))
blockchain_words = set(re.findall(r'\b[a-zA-Z]+\b', blockchain_text.lower()))

ai_words = ai_words - stop_words
blockchain_words = blockchain_words - stop_words

intersection = ai_words.intersection(blockchain_words)
union = ai_words.union(blockchain_words)

jaccard = len(intersection) / len(union)

print("\nJaccard Similarity:")
print(jaccard)

texts = [ai_text, blockchain_text]

vectorizer = TfidfVectorizer(stop_words='english')
vectors = vectorizer.fit_transform(texts)

cosine = cosine_similarity(vectors[0:1], vectors[1:2])

print("\nCosine Similarity:")
print(cosine[0][0])

print("\nAnalysis:")
print("Jaccard similarity compares common unique words.")
print("Cosine similarity compares TF-IDF vectors.")
print("Cosine similarity gives better insights when word importance matters.")


# Q5

print("\nQ5")

reviews = [
    "This product is amazing and works perfectly.",
    "I really love this product. It is worth buying.",
    "The quality is excellent and the service is great.",
    "This product is terrible and completely useless.",
    "I am disappointed with the poor quality.",
    "The product is okay and does its basic job."
]

analyzer = SentimentIntensityAnalyzer()

positive_reviews = []

for review in reviews:
    blob = TextBlob(review)
    vader = analyzer.polarity_scores(review)

    print("\nReview:", review)
    print("TextBlob Polarity:", blob.sentiment.polarity)
    print("TextBlob Subjectivity:", blob.sentiment.subjectivity)
    print("VADER Scores:", vader)

    if vader['compound'] >= 0.05:
        print("Sentiment: Positive")
        positive_reviews.append(review)

    elif vader['compound'] <= -0.05:
        print("Sentiment: Negative")

    else:
        print("Sentiment: Neutral")


if positive_reviews:
    text = " ".join(positive_reviews)

    wordcloud = WordCloud(
        width=800,
        height=400,
        background_color='white'
    ).generate(text)

    plt.figure(figsize=(10, 5))
    plt.imshow(wordcloud, interpolation='bilinear')
    plt.axis('off')
    plt.title("Positive Reviews Word Cloud")
    plt.show()


# Q6

print("\nQ6")

training_text = """
Technology plays an important role in modern life.
People use computers and smartphones every day.
Artificial intelligence helps machines learn from data.
Students use the internet to gain knowledge and improve skills.
Technology makes communication faster and businesses more efficient.
Machine learning helps computers identify patterns and make predictions.
New innovations continue to change the world.
"""

tokenizer = Tokenizer()
tokenizer.fit_on_texts([training_text])

total_words = len(tokenizer.word_index) + 1

sequences = []

for line in training_text.split('\n'):
    tokens = tokenizer.texts_to_sequences([line])[0]

    for i in range(1, len(tokens)):
        sequences.append(tokens[:i + 1])

max_length = max(len(seq) for seq in sequences)

sequences = pad_sequences(
    sequences,
    maxlen=max_length,
    padding='pre'
)

X = sequences[:, :-1]
y = sequences[:, -1]

model = Sequential()

model.add(Embedding(total_words, 50, input_length=max_length - 1))
model.add(LSTM(100))
model.add(Dense(total_words, activation='softmax'))

model.compile(
    loss='sparse_categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

model.fit(X, y, epochs=150, verbose=0)

def generate_text(seed_text, next_words):
    for i in range(next_words):
        token_list = tokenizer.texts_to_sequences([seed_text])[0]
        token_list = pad_sequences(
            [token_list],
            maxlen=max_length - 1,
            padding='pre'
        )

        predicted = model.predict(token_list, verbose=0)
        predicted_word = np.argmax(predicted, axis=1)[0]

        output_word = ""

        for word, index in tokenizer.word_index.items():
            if index == predicted_word:
                output_word = word
                break

        if output_word == "":
            break

        seed_text += " " + output_word

    return seed_text


print("\nGenerated Text:")
print(generate_text("Technology", 10))
print(generate_text("Artificial intelligence", 10))
print(generate_text("Students use", 10))
```
